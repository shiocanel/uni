```python
from random import sample


def prob_cond(nr_sim):
    countA, countB = 0, 0

    for _ in range(nr_sim):
        bile_extrase = sample(['red','blue','green'],counts=[5,3,2],k=3)
        if 'red' in bile_extrase:
            countA += 1

            if len(set(bile_extrase)) == 1:
                countB += 1
    
    return countB / countA

prob_cond(5000)
```

```python
from random import randrange

from matplotlib.pyplot import bar, grid, hist, legend, show

data = [randrange(1,7) for _ in range(500)]
bin_edges = [k+0.5 for k in range(7)]

hist(data, bin_edges, density = True, rwidth = 0.9, color = 'green', edgecolor = 'black',alpha = 0.5, label = 'frecvente relative')

# frecventa relativa -> density = True

distribution = {i: 1/6 for i in range(1, 7)}
bar(list(distribution.keys()), list(distribution.values()), width = 0.85, color = 'red', edgecolor = 'black', alpha = 0.6, label = 'probabilitati')

legend(loc = 'lower left')
grid()
show()
```

```python
def extrage_suma():
    bile_extrase = sample([1, 0], counts=[6, 4], k=5)

    suma = sum(bile_extrase)
    return suma
```
