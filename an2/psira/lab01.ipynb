```python
from random import sample
import random
from math import perm, comb
from itertools import permutations, combinations
```

```python
def aranjamente(s, length, numar_total=False, aleator=False):
    if aleator:    
        return sample(s, length)
    
    li = list(permutations(s, length))

    if numar_total:
        return len(li)

    return li

def combinari(s, length, numar_total=False, aleator=False):
    if aleator:    
        idx = [i for i in range(len(s))]

        p = idx[:]
        random.shuffle(p)

        l = p[:]
        r = l[:length]

        r.sort()
        return [s[i] for i in r]
    
    li = list(combinations(s, length))

    if numar_total:
        return len(li)

    return li
```

```python
print(aranjamente('word', 2))
print(aranjamente('word', 2, numar_total=True))
print(aranjamente('word', 2, aleator=True))

print(combinari('word', 2))
print(combinari('word', 2, numar_total=True))
print(combinari('word', 2, aleator=True))
```

```python
#problema
from itertools import combinations_with_replacement
from math import factorial
```

```python
def solve(s, cate=False):
    if cate:
        #(n + k - 1)! / r!(n-1)!

        return factorial(len(s) + 3 - 1) / (factorial(3) * factorial(len(s) - 1))
    
    li = list(combinations_with_replacement(s, 3))
    return li


print(solve('ACFKMVZ'))
print(solve('ACFKMVZ', cate=True))
```
