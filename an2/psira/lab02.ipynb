```python
counts = [0.0] * 20
for _ in range(100000):
    x = randint(1, 6)
    y = randint(1, 6)
    z = randint(1, 6)

    counts[x + y + z] += 1

print(counts)
```

```python
from math import dist
from matplotlib.pyplot import axis, plot, show
from random import random
```

```python
axis('square')
axis((0, 1, 0, 1))

x = [random() for _ in range(10000)]
y = [random() for _ in range(10000)]
plot(x, y, 'ro')
show()
```

```python
counts = []
x_int = []
y_int = []

for x_i, y_i in zip(x, y):
    if dist((x_i, y_i), (0.5, 0.5)) <= 0.5:
        x_int.append(x_i)
        y_int.append(y_i)
        counts.append((x_i, y_i))

axis('square')
axis((0, 1, 0, 1))
plot(x_int, y_int, 'ro')
```

```python
print(len(counts) / (10000) * 4)
```

```python
A = [0, 0]
B = [0, 1]
C = [1, 1]
D = [1, 0]
```

```python
counts = []
x_int = []
y_int = []

for x_i, y_i in zip(x, y):
    if dist((x_i, y_i), (0.5, 0.5)) <= dist((x_i, y_i), A) and dist((x_i, y_i), (0.5, 0.5)) <= dist((x_i, y_i), B) and dist((x_i, y_i), (0.5, 0.5)) <= dist((x_i, y_i), C) and dist((x_i, y_i), (0.5, 0.5)) <= dist((x_i, y_i), D):
        x_int.append(x_i)
        y_int.append(y_i)
        counts.append((x_i, y_i))


axis('square')
axis((0, 1, 0, 1))
plot(x_int, y_int, 'ro')
```
