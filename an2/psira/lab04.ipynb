```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, bernoulli
```

```python
def sari(pasi):
    res = []
    
    pos = 0
    for _ in range(pasi):
        p = bernoulli.rvs(p=0.5)
        pos += (1 if p == 1 else -1)

        res.append(pos)

    return res
```

```python
res = sari(100)

print(res)
print(res[-1])
```

```python
plt.plot(res)
plt.xlabel("Step")
plt.ylabel("Position")
plt.grid(True)
plt.show()
```

```python
simulare = []
for _ in range(1000):
    simulare.append(sari(100))

print(simulare)
```

```python
for p in simulare:
    plt.plot(range(1, len(p) + 1), p, alpha=0.15)

plt.xlabel("Step")
plt.ylabel("Position")
plt.grid(True)
plt.show()
```

```python
# histograma

from random import randrange

from matplotlib.pyplot import bar, grid, hist, legend, show

```

```python
def sari2(pasi):
    res = []
    
    pos = 0
    for _ in range(pasi):
        p = bernoulli.rvs(p=0.5)
        pos += (1 if p == 1 else -1)

        res.append(pos)

    return res[-1]

```

```python
def simuleaza(N, pasi):
    res = []
    
    for _ in range(N):
        res.append(sari2(pasi))

    return res
```

```python
N = 1000
pasi = 11

simulare = simuleaza(N, pasi)
print(simulare)
```

```python
data = simulare

bin_edges = np.arange(-pasi - 0.5, pasi + 1.5, 1)
hist(data, bins=bin_edges.tolist(), density=True, color='black', alpha=0.5, label='frecventa relativa')

positions = np.arange(-pasi, pasi + 1, 2)
distribution = binom.pmf((positions + pasi) // 2, pasi, 0.5)
bar(positions, distribution, width=0.85, color='red', edgecolor='black', alpha=0.6, label='probabilitati')

legend(loc = 'lower left')
grid()
plt.xticks(np.arange(-pasi, pasi + 1, 2))
show()
```

```python
def simuleaza_cerc(N, pasi):
    res = []
    
    for _ in range(N):
        res.append(sari_cerc(pasi))

    return res
```
